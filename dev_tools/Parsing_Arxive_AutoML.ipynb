```python
import arxiv

client = arxiv.Client()

search = arxiv.Search(
    query=["AutoML", "Neural Architecture Search", "Hyperparameter Optimization", "AutoDL", 
          "NAS", "Full-Pipeline AutoML", "Automated Feature Engineering", "Feature Discovery", 
          "Automated Model Selection", "HPO", "Automated Model Training", "AutoML-Agent", 
          "Agentic AutoML", "Multi-Agent Framework for AutoML", "LLM Agents for Automated Machine Learning",
          "Memory-Augmented Multi-Agent System", "Tree-Search Enhanced LLM Agents", 
          "Auto Research System / Autonomous Model Discovery", "Machine Learning Testing (ML Testing)",
          "Automated Model Validation", "Verification and Validation (V&V) of ML Models", 
           "Automated Compliance Verification", "Fuzz Testing for ML / Deep Learning Fuzzing", 
          "Test Case Generation for ML", "Metamorphic Testing", "Automated Oracle for ML Testing", 
          "Robustness / Fairness Testing", "MLOps (Machine Learning Operations)", "Automated ML Pipeline / Workflow",
          "Automated Model Deployment", "Continuous Delivery for Machine Learning (CD4ML)", 
           "Agent Decisions Auditing", "Sandboxed Reflection Loop"],
    max_results=100,
    sort_by=arxiv.SortCriterion.SubmittedDate,
)

for r in client.results(search):
    print("=" * 80)
    print("1. Название:", r.title)

    print("2. Авторы:", ", ".join(a.name for a in r.authors))

    # Аннотация приходит с переносами строк — нормализуем пробелы
    abstract = " ".join(r.summary.split())
    print("3. Аннотация:", abstract)

    # Дата подачи (обычно это и есть год публикации препринта)
    print("4. Дата подачи:", r.published.strftime("%Y-%m-%d"))
    print("   Обновлено:", r.updated.strftime("%Y-%m-%d"))

    # Параметры издания / метаданные arXiv
    print("   arXiv ID:", r.get_short_id())
    print("   Версия:", r.get_short_id().split("v")[-1])
    print("   Основная категория:", r.primary_category)
    print("   Все категории:", ", ".join(r.categories))
    print("   DOI:", r.doi or "—")
    print("   Journal ref:", r.journal_ref or "—")
    print("   Comment:", r.comment or "—")
    print("   PDF:", r.pdf_url)
```
