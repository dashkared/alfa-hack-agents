### A. Классификация (1–25)

1. **Accuracy** — доля всех правильных предсказаний.
2. **Balanced Accuracy** — accuracy со сбалансированным весом классов (полезна при дисбалансе).
3. **Precision** — доля истинно положительных среди всех предсказанных положительных.
4. **Recall (Sensitivity)** — доля истинно положительных среди всех реальных положительных.
5. **Specificity** — доля истинно отрицательных среди всех реальных отрицательных.
6. **F1-Score** — гармоническое среднее Precision и Recall.
7. **F-beta Score** — обобщение F1 с весом β (β>1 — акцент на Recall, β<1 — на Precision).
8. **Matthews Correlation Coefficient (MCC)** — корреляция между предсказанием и истиной, устойчива к дисбалансу.
9. **Cohen’s Kappa** — согласие с истиной, скорректированное на случайное совпадение.
10. **ROC-AUC** — площадь под ROC-кривой (TPR vs FPR).
11. **PR-AUC / Average Precision (AP)** — площадь под кривой Precision-Recall, чувствительна к миноритарному классу.
12. **Log-Loss (Cross-Entropy)** — штраф за уверенные неверные вероятностные прогнозы.
13. **Brier Score** — среднеквадратичная ошибка между предсказанной вероятностью и фактическим исходом.
14. **Hinge Loss** — функция потерь для SVM-подобных моделей (максимальный зазор).
15. **Hamming Loss** — доля неверно предсказанных меток (для мультилейбл).
16. **Jaccard Score** — пересечение над объединением множеств истинных и предсказанных меток.
17. **Zero-One Loss** — доля ошибочных предсказаний (1 — ошибка, 0 — верно).
18. **Confusion Matrix (raw counts)** — матрица TP/TN/FP/FN как исходный артефакт для остальных метрик.
19. **Classification Report (per-class)** — precision/recall/F1/support по каждому классу.
20. **Top-K Accuracy** — доля случаев, когда истинный класс попал в K наиболее вероятных предсказаний.
21. **Lift@K** — во сколько раз модель чаще находит положительный класс в топ-K, чем случайный выбор.
22. **Class Likelihood Ratios** — отношения правдоподобий для положительного и отрицательного результатов теста.
23. **G-mean** — геометрическое среднее Sensitivity и Specificity (устойчива к дисбалансу).
24. **Optimized Threshold (G-means)** — значение порога, максимизирующее G-mean, а не фиксированное 0.5.
25. **Expected Calibration Error (ECE)** — среднее расхождение между предсказанной уверенностью и фактической точностью (калибровка).

### B. Регрессия (26–48)

26. **Mean Squared Error (MSE)** — среднее квадратов ошибок; сильно штрафует выбросы.
27. **Root Mean Squared Error (RMSE)** — корень из MSE, ошибка в исходных единицах.
28. **Mean Absolute Error (MAE)** — среднее абсолютных ошибок; устойчивее к выбросам, чем MSE.
29. **Median Absolute Error (MedAE)** — медиана абсолютных ошибок, ещё более робастна к выбросам.
30. **Max Error** — максимальная абсолютная ошибка на наблюдении.
31. **R² (Coefficient of Determination)** — доля объяснённой дисперсии целевой переменной.
32. **Adjusted R²** — R² со штрафом за число признаков (для сравнения моделей разной сложности).
33. **Explained Variance Score** — доля дисперсии, которую модель объясняет (отличается от R² при смещении).
34. **Mean Absolute Percentage Error (MAPE)** — средняя абсолютная ошибка в процентах от фактического значения.
35. **Symmetric MAPE (SMAPE)** — симметричная версия MAPE, устойчива к нулевым значениям.
36. **Mean Squared Log Error (MSLE)** — MSE на логарифмах, штрафует недооценку сильнее переоценки.
37. **Root Mean Squared Log Error (RMSLE)** — корень из MSLE, интерпретируемая в лог-единицах.
38. **Huber Loss** — комбинация MSE и MAE; квадратична для малых ошибок, линейна для больших.
39. **Quantile (Pinball) Loss** — потеря для квантильной регрессии (например, для 90-го перцентиля).
40. **Tweedie Deviance** — обобщённая ошибка для распределений из семейства Tweedie (страховые/счётные данные).
41. **Relative Absolute Error (RAE)** — MAE модели, нормированная на MAE наивного прогноза.
42. **Relative Squared Error (RSE)** — MSE модели, нормированная на MSE наивного прогноза.
43. **Normalized RMSE (NRMSE)** — RMSE, делённый на размах или стандартное отклонение таргета.
44. **Interval Score** — оценка прогнозных интервалов (ширина + штраф за непопадание).
45. **Coverage Probability** — доля случаев, когда истинное значение попало в предсказанный интервал.
46. **Interval Width** — средняя ширина прогнозного интервала (уже — лучше при том же покрытии).
47. **Winkler Score** — комбинированная метрика ширины интервала и покрытия.
48. **Mean Absolute Scaled Error (MASE)** — MAE, масштабированная на MAE сезонного наивного прогноза (для временных рядов).

### C. Кластеризация: внутренние метрики (49–60)

49. **Silhouette Score** — насколько объект ближе к своему кластеру, чем к чужому (−1…+1).
50. **Silhouette Samples** — silhouette-значение для каждого отдельного объекта (для диагностики).
51. **Calinski-Harabasz Index** — отношение межкластерной дисперсии к внутрикластерной (выше — лучше).
52. **Davies-Bouldin Index** — среднее сходство между кластерами (ниже — лучше).
53. **Dunn Index** — минимальное межкластерное расстояние, делённое на максимальный диаметр кластера (выше — лучше).
54. **Inertia (Within-Cluster Sum of Squares)** — сумма квадратов расстояний до центроидов (для K-means).
55. **Gap Statistic** — сравнение inertia с ожидаемой при случайных данных (для выбора K).
56. **Cluster Stability** — устойчивость разбиения при ресемплинге данных.
57. **Consensus Scoring** — согласованность нескольких запусков кластеризации.
58. **Elbow Method (visual criterion)** — эвристика выбора K по «излому» кривой inertia.
59. **BIC (Bayesian Information Criterion)** — штраф за сложность модели кластеризации (для выбора K).
60. **AIC (Akaike Information Criterion)** — альтернативный информационный критерий для выбора модели.

---

### D. Кластеризация: внешние метрики (61–70)

61. **Adjusted Rand Index (ARI)** — сходство разбиений, скорректированное на случайность.
62. **Rand Index (RI)** — доля пар объектов, по которым разбиения совпадают.
63. **Normalized Mutual Information (NMI)** — взаимная информация между разбиениями, нормированная.
64. **Adjusted Mutual Information (AMI)** — NMI со штрафом за случайные совпадения.
65. **V-measure** — гармоническое среднее homogeneity и completeness.
66. **Homogeneity** — каждый кластер содержит объекты только одного класса.
67. **Completeness** — все объекты одного класса попали в один кластер.
68. **Fowlkes-Mallows Score** — геометрическое среднее попарной precision и recall.
69. **Contingency Matrix** — таблица сопряжённости «кластер × истинный класс».
70. **Pair Confusion Matrix** — матрица TP/FP/FN/TN на уровне пар объектов для кластеризации.

### E. Ранжирование и информационный поиск (71–82)

71. **Precision@K (P@K)** — доля релевантных документов среди первых K.
72. **Recall@K** — доля найденных релевантных документов среди всех релевантных в топ-K.
73. **Mean Average Precision (MAP)** — среднее AP по всем запросам.
74. **MAP@K** — MAP, ограниченный первыми K позициями.
75. **Mean Reciprocal Rank (MRR)** — среднее обратных рангов первого релевантного документа.
76. **MRR@K** — MRR в пределах первых K позиций.
77. **Normalized Discounted Cumulative Gain (NDCG)** — качество ранжирования с дисконтированием нижних позиций.
78. **NDCG@K** — NDCG, ограниченный топ-K.
79. **DCG (Discounted Cumulative Gain)** — сырая (ненормированная) версия NDCG.
80. **Kendall’s Tau** — ранговая корреляция между предсказанным и истинным порядком.
81. **Spearman’s Rank Correlation** — монотонная ранговая корреляция.
82. **Label Ranking Average Precision (LRAP)** — средняя точность ранжирования меток для мультилейбл-задач.

### F. Детекция объектов и сегментация (83–92)

83. **Intersection over Union (IoU)** — площадь пересечения над объединением для бокса или маски.
84. **Mean IoU (mIoU)** — среднее IoU по всем классам сегментации.
85. **Pixel Accuracy** — доля правильно классифицированных пикселей.
86. **Mean Pixel Accuracy** — среднее pixel accuracy по классам.
87. **Dice Coefficient** — удвоенная площадь пересечения над суммой площадей (F1 для масок).
88. **Jaccard Index (для масок)** — то же, что IoU, для сегментационных масок.
89. **Average Precision (AP) для детекции** — площадь под PR-кривой для одного класса при фиксированном IoU-пороге.
90. **mAP@50** — mean AP при IoU-пороге 0.5.
91. **mAP@50–95** — mean AP, усреднённый по IoU-порогам от 0.5 до 0.95 с шагом 0.05.
92. **Boundary F-measure** — F-мера по границам сегментации (а не по площади).

### G. Генеративные модели (93–97)

93. **Fréchet Inception Distance (FID)** — расстояние между распределениями признаков реальных и сгенерированных изображений (ниже — лучше).
94. **Inception Score (IS)** — качество и разнообразие сгенерированных изображений (выше — лучше).
95. **Kernel Inception Distance (KID)** — MMD между признаками Inception; несмещённая альтернатива FID.
96. **Maximum Mean Discrepancy (MMD)** — расстояние между распределениями в пространстве признаков.
97. **CMMD (CLIP-MMD)** — MMD на CLIP-признаках, лучше согласуется с человеческими оценками.

---

### H. Справедливость и специализированные (98–100)

98. **Demographic Parity Difference / Ratio** — разница (или отношение) долей положительных исходов между защищёнными группами.
99. **Equalized Odds Difference** — разница TPR и FPR между привилегированной и непривилегированной группами.
100. **Disparate Impact Ratio** — отношение долей положительных исходов между группами (порог 0.8 по правилу «four-fifths»).

### A. Прогнозирование временных рядов (1–15)

1. **CRPS (Continuous Ranked Probability Score)** — proper scoring rule для вероятностных прогнозов; обобщает MAE на распределения.
2. **CRPS-sum** — CRPS для агрегированного/многомерного прогноза.
3. **Weighted Interval Score (WIS)** — комбинирует ширину интервалов и покрытие по набору квантилей.
4. **Mean Scaled Interval Score (MSIS)** — WIS, масштабированный на ошибку сезонного наивного прогноза.
5. **Energy Score** — многомерный proper score для совместных распределений.
6. **Variogram Score** — многомерный score, чувствительный к структуре зависимостей между рядами.
7. **Mean Pinball Loss (агрегированный по квантилям)** — средняя квантильная потеря по всем уровням квантилей.
8. **Forecast Bias (MPE, Mean Percentage Error)** — систематическое переоценивание/недооценивание прогноза.
9. **Tracking Signal** — накопленное смещение, делённое на MAD; индикатор «уплывания» прогноза.
10. **Theil's U statistic** — отношение ошибки модели к ошибке наивного прогноза.
11. **Forecast Value Added (FVA)** — прирост качества относительно бенчмарка (naive/seasonal naive).
12. **Diebold-Mariano statistic** — статистическая значимость различия в точности двух прогнозов.
13. **Ljung-Box Q statistic** — автокорреляция остатков (проверка, что модель «выпила» всю структуру).
14. **Durbin-Watson statistic** — автокорреляция остатков первого порядка.
15. **Dynamic Time Warping (DTW) distance** — расстояние выравнивания последовательностей, устойчиво к сдвигам во времени.

### B. Детекция аномалий (16–25)

16. **Range-based Precision** — доля предсказанных аномальных интервалов, пересекающихся с истинными.
17. **Range-based Recall** — доля истинных аномальных интервалов, которые модель обнаружила.
18. **Point-Adjusted Precision** — precision, где попадание хотя бы в одну точку интервала засчитывается как полное обнаружение.
19. **Point-Adjusted Recall** — аналогично для recall.
20. **Point-Adjusted F1** — F1 на основе point-adjusted precision/recall.
21. **Affiliation F1** — F1 с толерантностью к частичному перекрытию интервалов (без «всё или ничего»).
22. **Detection Delay** — среднее число шагов между началом истинной аномалии и первым срабатыванием.
23. **False Alarm Rate (FAR)** — число ложных срабатываний на единицу времени / на число нормальных точек.
24. **Detection Rate (DR)** — доля обнаруженных аномалий от всех истинных.
25. **Volume Under ROC Surface (VUS)** — обобщение AUC для range-based детекции аномалий.

### C. Анализ выживаемости (26–33)

26. **Harrell's Concordance Index (C-index)** — доля корректно упорядоченных пар по времени выживания.
27. **Uno's C-index** — C-index со поправкой на цензурирование.
28. **Time-dependent AUC(t)** — дискриминация модели в конкретный момент времени t.
29. **Integrated AUC (iAUC)** — AUC, проинтегрированный по времени.
30. **Cumulative/Dynamic AUC** — альтернативная time-dependent AUC с иным определением контролов.
31. **Integrated Brier Score (IBS)** — Brier score, проинтегрированный по времени (калибровка + дискриминация).
32. **Brier Score at time t** — квадратичная ошибка предсказанной вероятности выживания в момент t.
33. **RMST difference (Restricted Mean Survival Time)** — разница ожидаемого времени жизни до горизонта τ.

### D. Рекомендательные системы (34–45)

34. **Hit Rate@K** — доля пользователей, у которых хотя бы один релевантный объект попал в топ-K.
35. **Catalog Coverage** — доля объектов каталога, которые хоть раз были рекомендованы.
36. **User Coverage** — доля пользователей, получивших непустые рекомендации.
37. **Intra-List Diversity (ILD)** — средняя попарная несхожесть объектов внутри одной выдачи.
38. **Novelty** — средняя «неожиданность» рекомендованных объектов (self-information).
39. **Serendipity** — неожиданные, но релевантные рекомендации.
40. **Personalization** — насколько выдачи разных пользователей непохожи друг на друга.
41. **Popularity Bias** — расхождение между распределением рекомендаций и популярностью объектов (Gini/KL).
42. **Average Recommendation Popularity (ARP)** — средняя популярность рекомендованных объектов.
43. **Expected Popularity Complement (EPC)** — насколько хорошо модель «достаёт» длинный хвост.
44. **GAUC (Group AUC)** — AUC, усреднённый по пользователям с весом по числу показов.
45. **Cold-Start Hit Rate** — hit rate, ограниченный холодными объектами/пользователями.

### E. NLP и генерация текста (46–60)

46. **BLEU** — n-gram precision с brevity penalty.
47. **SacreBLEU** — воспроизводимый вариант BLEU с фиксированной токенизацией.
48. **ROUGE-N** — n-gram recall (ROUGE-1, ROUGE-2) для суммаризации.
49. **ROUGE-L** — recall на основе наибольшей общей подпоследовательности.
50. **METEOR** — precision/recall с учётом синонимов и стемминга.
51. **chrF** — character n-gram F-score, устойчив к морфологии.
52. **TER (Translation Edit Rate)** — нормированное расстояние редактирования до эталона.
53. **BERTScore** — семантическое сходство через контекстные эмбеддинги.
54. **MoverScore** — Earth Mover's Distance между эмбеддингами текстов.
55. **BLEURT** — обученная метрика, согласованная с человеческими оценками.
56. **COMET** — обученная метрика качества машинного перевода.
57. **Exact Match (EM)** — точное совпадение ответа с эталоном (QA).
58. **Token-level F1 (SQuAD)** — F1 по токенам между предсказанным и эталонным ответом.
59. **Perplexity** — экспонента средней отрицательной лог-вероятности (качество языковой модели).
60. **MAUVE** — расхождение между распределениями сгенерированного и человеческого текста.

### F. Речь и аудио (61–68)

61. **Word Error Rate (WER)** — доля ошибок на уровне слов в распознавании речи.
62. **Character Error Rate (CER)** — доля ошибок на уровне символов.
63. **Sentence Error Rate (SER)** — доля предложений с хотя бы одной ошибкой.
64. **Match Error Rate (MER)** — ошибка с учётом совпадений, альтернатива WER.
65. **PESQ** — перцептивная оценка качества речи.
66. **STOI** — кратковременная объективная разборчивость речи.
67. **SI-SDR** — масштабно-инвариантное отношение сигнал/искажение (разделение источников).
68. **MOS / UTMOS** — субъективная (или обученная) оценка качества звучания.

### G. Multi-label и ordinal-задачи (69–74)

69. **Micro-F1** — F1, посчитанный по всем парам «объект-метка» сразу.
70. **Macro-F1** — невзвешенное среднее F1 по меткам (все метки равны).
71. **Weighted-F1** — F1, взвешенный по поддержке (support) каждой метки.
72. **Subset Accuracy (Exact Match Ratio)** — доля объектов, у которых все метки предсказаны верно.
73. **Label Ranking Loss** — средняя доля неправильно упорядоченных пар меток.
74. **Quadratic Weighted Kappa (QWK)** — согласие для ordinal-предсказаний со штрафом, растущим квадратично от расстояния между классами.

### H. Снижение размерности и представления (75–80)

75. **Trustworthiness** — насколько соседи в низкоразмерном пространстве действительно были соседями в исходном.
76. **Continuity** — насколько соседи из исходного пространства сохранились в проекции.
77. **Reconstruction Error** — MSE между исходными данными и восстановленными из латентного пространства.
78. **Explained Variance Ratio** — доля дисперсии, объяснённая компонентами (PCA).
79. **kNN Preservation** — доля совпадающих k-ближайших соседей до и после проекции.
80. **Shepard Correlation** — корреляция между попарными расстояниями в исходном и вложенном пространствах.

### I. Обучение с подкреплением (81–88)

81. **Average Return** — средняя накопленная награда за эпизод.
82. **Sample Efficiency** — return на единицу взаимодействий со средой.
83. **Regret** — накопленная разница между оптимальной и фактической политикой.
84. **Success Rate** — доля эпизодов, в которых достигнута цель.
85. **Episode Length** — среднее число шагов до завершения эпизода.
86. **Value Function MSE** — ошибка оценки функции ценности относительно реальных return'ов.
87. **Advantage Estimation Error** — ошибка оценки advantage-функции.
88. **Policy Entropy** — энтропия политики (мера исследования/детерминированности).

### J. Неопределённость и калибровка (89–93)

89. **Negative Log-Likelihood (NLL)** — средняя отрицательная лог-вероятность предсказательного распределения.
90. **Sharpness** — средняя предсказательная дисперсия (узость распределения).
91. **Calibration Slope** — наклон регрессии наблюдаемых частот на предсказанные вероятности.
92. **Calibration Intercept** — свободный член той же регрессии (систематический сдвиг калибровки).
93. **Spread-Skill Ratio** — отношение средней предсказательной σ к фактической ошибке; >1 — модель «неуверенна», <1 — «самоуверенна».

---

### K. Робастность и состязательность (94–98)

94. **Adversarial Accuracy** — accuracy на состязательно возмущённых входах.
95. **Attack Success Rate** — доля успешных состязательных атак.
96. **Robustness Gap** — разница между clean accuracy и accuracy на возмущённых данных.
97. **Certified Robustness Radius** — гарантированный радиус возмущения, в пределах которого предсказание не меняется.
98. **Corruption Error (mCE)** — относительная ошибка на коррумпированных данных (ImageNet-C стиль), нормированная на baseline.

---

### L. Эффективность (99–100)

99. **Inference Latency** — время на один пример (или батч) при инференсе.
100. **Throughput** — число обработанных примеров в секунду (сэмплов/сек).



### 🕸️ Графовые нейронные сети (GNN) (1–18)

Метрики GNN сильно зависят от задачи. Для **классификации узлов** — стандартный набор классификации; для **предсказания связей** — ранжирующие метрики; для **графовой регрессии** — ошибки регрессии.

1. **Node Classification Accuracy** — доля верно классифицированных узлов.
2. **Node Classification F1 (macro/micro)** — F1 по узлам, macro или micro в зависимости от баланса классов.
3. **Node Classification ROC-AUC** — площадь под ROC-кривой для узлов (мультикласс — one-vs-rest).
4. **Link Prediction Hits@K** — попал ли истинный положительный объект в топ-K предсказанных связей.
5. **Link Prediction MRR (Mean Reciprocal Rank)** — средний обратный ранг первого релевантного объекта.
6. **Link Prediction AUC** — вероятность, что положительная связь ранжирована выше случайной отрицательной.
7. **Link Prediction Average Precision (AP)** — площадь под PR-кривой для предсказания связей.
8. **Graph Classification Accuracy** — доля верно классифицированных графов.
9. **Graph Regression MAE / RMSE** — ошибки регрессии на уровне графа.
10. **Node Clustering NMI** — нормализованная взаимная информация между предсказанными и истинными кластерами узлов.
11. **Node Clustering ARI (Adjusted Rand Index)** — скорректированное на случайность согласие кластеризации.
12. **Recall@K для рекомендаций на графах** — доля релевантных объектов в топ-K рекомендаций.
13. **NDCG@K для рекомендаций на графах** — качество ранжирования с дисконтированием.
14. **Hit Ratio** — доля случаев, когда хотя бы один релевантный объект попал в топ-K.
15. **Robustness to Noisy Graph Structures** — деградация метрик при добавлении шума в рёбра/признаки.
16. **Scalability (время на эпоху / память)** — практическая метрика для больших графов.
17. **Fairness across Node Groups** — разрыв в метриках между группами узлов (по защищённому атрибуту).
18. **Explanation Fidelity для GNN** — насколько объяснение (например, edge mask) верно отражает решение модели.

---

### 🏛️ Федеративное обучение (19–30)

Помимо стандартных accuracy/precision/recall/F1, федеративное обучение добавляет метрики **приватности**, **коммуникационной эффективности** и **устойчивости к не-IID данным**.

19. **Global Model Test Accuracy** — accuracy глобальной модели на централизованном тесте.
20. **Global Model F1 / Precision / Recall** — стандартные метрики для глобальной модели.
21. **Average Train Accuracy per Client** — средняя локальная accuracy по клиентам.
22. **Average Train Loss per Client** — средняя локальная потеря.
23. **Accuracy Loss / Accuracy Drop** — падение accuracy относительно централизованного baseline.
24. **Privacy Leakage Rate** — количественная оценка утечки приватности (например, через membership inference).
25. **Communication Cost** — объём переданных данных за раунд обучения.
26. **Average Training Time per Round** — время одного раунда агрегации.
27. **Non-IIDness Metric (label skew)** — метрика степени неоднородности распределения меток по клиентам.
28. **Marginal Contribution of a Client** — вклад конкретного клиента в глобальную модель.
29. **Generalization Gap** — разрыв между тренировочной и валидационной метрикой на клиентах.
30. **Fairness across Clients** — дисперсия метрик между клиентами (не должно быть «слабых» клиентов).

---

### 📦 Сжатие и квантование моделей (31–40)

Здесь ключевое — **сохранить качество при сокращении размера / ускорении**. Поэтому метрики парные: качество + эффективность.

31. **Model Size (KB / MB)** — размер сжатой модели.
32. **Compression Ratio** — отношение размера исходной модели к сжатой.
33. **Test Accuracy (after compression)** — accuracy после квантования/прунинга.
34. **Accuracy Drop (percentage points)** — падение accuracy относительно FP32-baseline.
35. **Test Inference Time (ms)** — время инференса на тестовом батче.
36. **Real-time Inference Time (ms)** — задержка в реальном времени.
37. **Throughput (samples/sec)** — пропускная способность.
38. **Perplexity (для LLM после квантования)** — качество языковой модели в сжатом виде.
39. **mAP@0.5 / mAP@0.5:0.95 (для детекции после сжатия)** — сохранение качества детекции.
40. **FLOPs Reduction** — сокращение вычислительной сложности.

---

### 🌓 Semi-supervised learning (41–48)

В semi-supervised сценарии метрики те же, что и в supervise, но **при фиксированном малом проценте меток** — иначе сравнение нечестное.

41. **Accuracy at Labeled Fraction p%** — accuracy при доле размеченных данных `p` (например, 10%).
42. **Precision / Recall / F1 at p%** — стандартный набор при ограниченной разметке.
43. **mIoU (для сегментации)** — mean IoU при semi-supervised сегментации.
44. **Dice Coefficient (DSC)** — коэффициент Дайса для медицинской сегментации.
45. **Aggregated Jaccard Index (AJI)** — агрегированный Jaccard для instance-сегментации.
46. **Object-level F1 (F1obj)** — F1 на уровне объектов, а не пикселей.
47. **Pseudo-label Quality** — точность псевдо-меток на валидации.
48. **Label Efficiency Curve** — площадь под кривой «качество vs. доля меток».

---

### 🎯 Active learning (49–56)

В active learning важно не только **итоговое качество**, но и **сколько меток потребовалось** для его достижения.

49. **Accuracy vs. Number of Labels** — кривая зависимости accuracy от числа запрошенных меток.
50. **Area Under Learning Curve (ALC)** — площадь под кривой «качество vs. бюджет разметки».
51. **Sample Efficiency** — прирост качества на одну размеченную метку.
52. **Achievable Accuracy (A_max)** — асимптотическое качество, к которому сходится стратегия.
53. **Coverage Efficiency (δ)** — насколько быстро стратегия покрывает пространство данных.
54. **Early-stage Performance (α)** — качество на первых раундах разметки.
55. **Rank-to-Positive (RTP)** — позиция первого релевантного документа в ранжировании (для retrieval).
56. **Agreement Rate** — доля меток, с которыми согласился бы «оракул» (человек/LLM).

---

### 🔗 Причинно-следственный вывод (Causal Inference) (57–66)

Метрики делятся на две группы: **точность восстановления причинного графа** и **точность оценки эффектов** (ITE/ATE).

57. **PEHE (Precision in Estimation of Heterogeneous Effect)** — √MSE между оценённым и истинным индивидуальным эффектом лечения.
58. **ATE Error (ε_ATE)** — абсолютная ошибка среднего эффекта лечения.
59. **Policy Risk** — риск решения, принятого на основе оценённых ITE.
60. **Causal Graph Precision / Recall / F1** — точность восстановления рёбер причинного графа относительно ground truth.
61. **Structural Hamming Distance (SHD)** — число рёбер, которые нужно добавить/удалить/развернуть, чтобы получить истинный граф.
62. **False Discovery Rate (FDR) причинных связей** — доля ложных причинных связей среди обнаруженных.
63. **Generalization Error for Causal Models** — способность модели обобщаться на новые подгруппы/среды.
64. **Surrogacy Metric (ICA / Rényi-based)** — качество суррогатной переменной для оценки эффекта лечения.
65. **Counterfactual Calibration** — калибровка предсказанных counterfactual-исходов.
66. **Covariate Balance (ASMD)** — стандартизованная разница средних ковариат до/после взвешивания (для propensity score методов).

---

### 🌊 Дрейф концепции (Concept Drift) (67–76)

Метрики дрейфа оценивают **детектор**, а не модель: насколько быстро и точно он ловит сдвиг распределения.

67. **F1 Detection Score** — гармоническое среднее precision и recall детектора дрейфа.
68. **Drift Recall** — доля реальных дрейфов, которые детектор обнаружил.
69. **False Alarm Rate (FAR)** — число ложных срабатываний на единицу времени.
70. **Alarm Rate (AR)** — общее число тревог на единицу времени.
71. **Normalized Detection Time (NDT)** — задержка обнаружения, нормированная на максимально допустимую.
72. **Balanced Accuracy by Requested Labels (BAR)** — компромисс между accuracy детектора и долей запрошенных меток.
73. **KL Drift Intensity** — сила дрейфа через дивергенцию Кульбака–Лейблера между окнами.
74. **Conditional Entropy Drift** — разница условных энтропий между временными окнами.
75. **Accuracy over Rolling Windows** — падение accuracy модели после дрейфа (performance-based monitoring).
76. **Calibration Error Drift** — изменение ECE после дрейфа.

---

### 🔍 Интерпретируемость (XAI) (77–86)

Метрики XAI оценивают **качество объяснений**, а не качество модели. Основные свойства: fidelity, stability, complexity, comprehensiveness.

77. **Faithfulness / Fidelity** — насколько объяснение верно отражает реальное рассуждение модели.
78. **Stability / Robustness** — согласованность объяснений при малых возмущениях входа.
79. **Complexity** — сложность объяснения (число правил, признаков, глубина дерева).
80. **Consistency** — согласованность объяснений для похожих входов.
81. **Comprehensiveness** — насколько удаление «важных» признаков меняет предсказание.
82. **Sufficiency** — достаточно ли «важных» признаков для сохранения предсказания.
83. **Precision of Explanation** — доля действительно важных признаков среди выделенных.
84. **Coverage of Explanation** — доля случаев, для которых объяснение вообще может быть построено.
85. **Explainability-Performance Coefficient (EPC)** — компромисс между интерпретируемостью и качеством модели.
86. **Causal Alignment** — согласованность объяснения с причинными связями в данных.

---

### 🧩 Мета-обучение и few-shot (87–94)

87. **N-way K-shot Accuracy** — accuracy на новых классах после `K` примеров каждого из `N` классов.
88. **Meta-test Accuracy** — accuracy на мета-тестовом наборе задач.
89. **Adaptation Speed** — число шагов/примеров до достижения целевого качества.
90. **Generalization Gap (meta-train vs meta-test)** — переобучение на мета-уровне.
91. **Catastrophic Forgetting Rate** — падение качества на старых задачах после изучения новых.
92. **Forward Transfer** — положительный перенос от предыдущих задач к новым.
93. **Backward Transfer** — улучшение старых задач после изучения новых.
94. **Sample Efficiency (meta)** — сколько примеров нужно для достижения целевого качества на новой задаче.

---

### ⚖️ Несбалансированное обучение (95–100)

Эти метрики уже частично пересекались с классификацией, но здесь — **специфичный фокус на дисбалансе**.

95. **Minority Class Recall** — recall именно миноритарного класса.
96. **Minority Class F1** — F1 миноритарного класса.
97. **G-mean (Geometric Mean of Sens/Spec)** — геометрическое среднее чувствительности и специфичности.
98. **Balanced Accuracy** — accuracy со сбалансированным весом классов.
99. **MCC (Matthews Correlation Coefficient)** — корреляция, устойчивая к дисбалансу.
100. **PR-AUC (Precision-Recall AUC)** — площадь под PR-кривой, чувствительная к миноритарному классу.
